```python
import pandas as pd
from datetime import datetime
import pickle

filepath = "data/Games.csv"

pd.set_option('display.max_columns', None)

```

```python
df_games = pd.read_csv(filepath)

df_games.info()
```

```python
df_games['gameDate'] = pd.to_datetime(df_games['gameDateTimeEst']).dt.date
df_games.sort_values(by='gameDate', inplace=True)
df_games.reset_index(drop=True, inplace=True)
```

```python
df_games.isna().sum()
```

```python
df_games.drop(df_games.iloc[:, [12, 14, 18, 19, 20, 21]], axis=1, inplace=True)
```

```python
df_games.to_csv("data/Games.csv", index=False)
```

```python

df_games["year"] = pd.to_datetime(df_games["gameDateTimeEst"]).dt.year
df_games["month"] = pd.to_datetime(df_games["gameDateTimeEst"]).dt.month
df_games["day"] = pd.to_datetime(df_games["gameDateTimeEst"]).dt.day
df_games["DoW"] = pd.to_datetime(df_games["gameDateTimeEst"]).dt.day_of_week
df_games["time"] = pd.to_datetime(df_games["gameDateTimeEst"]).dt.time
df_games.head(3)
```

```python
df_games_playoff_2023 = df_games[(df_games["gameType"] == "Playoffs") & (df_games["year"] == 2023)]
#df_games_playoff_2023.drop(columns=['attendance', 'arenaId','arenaName', 'arenaCity', 'arenaState', 'officials', 'year', 'month',
#       'day', 'DoW', 'time'], inplace=True)
```

```python
df_games_playoff_2023
```

```python
df_games_west_2023 = df_games_playoff_2023[(df_games_playoff_2023["gameLabel"] == "West - First Round") & (df_games_playoff_2023["seriesGameNumber"] == 1)]
df_games_east_2023 = df_games_playoff_2023[(df_games_playoff_2023["gameLabel"] == "East - First Round") & (df_games_playoff_2023["seriesGameNumber"] == 1)]
df_games_west_2023
```

```python

df_games_boston_2023 = df_games_playoff_2023[(df_games_playoff_2023["hometeamName"] == "Celtics") | (df_games_playoff_2023["awayteamName"] == "Celtics")]
df_games_boston_2023_1 = df_games_boston_2023[df_games_boston_2023["seriesGameNumber"] == 1]
df_games_boston_2023_1.drop(df_games_boston_2023_1.iloc[:, [12, 14, 18, 19, 20, 21]], axis=1, inplace=True)
```
